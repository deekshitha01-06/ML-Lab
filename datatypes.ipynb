```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

```python
data = pd.read_csv("hospital_data.csv")

print("Original Dataset")
print(data)
```

```python
# 2. Identify data types
print("\nData Types")
print(data.dtypes)
```

```python
# 3. Separate numerical and categorical attributes
numerical = data.select_dtypes(include=['int64', 'float64'])
categorical = data.select_dtypes(include=['object'])

print("\nNumerical Attributes")
print(numerical.columns)

print("\nCategorical Attributes")
print(categorical.columns)
```

```python
# 4. Convert categorical variables into numerical form
encoder = LabelEncoder()

for column in categorical.columns:
    data[column] = encoder.fit_transform(data[column])

print("\nDataset After Encoding")
print(data)
```

```python
# 5. Remove unnecessary columns
data = data.drop(columns=["PatientID", "Name"])

print("\nDataset After Removing Unnecessary Columns")
print(data)
```

```python
# 6. Display summary statistics
print("\nSummary Statistics")
print(data.describe())
```

```python
data.to_csv("cleaned_hospital_data.csv", index=False)

print("\nCleaned dataset saved successfully.")
```
