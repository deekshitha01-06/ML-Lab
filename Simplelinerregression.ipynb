```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
data = {'Hours':[1, 2, 3, 4, 5],
        'marks':[2, 3, 5, 4, 6]}
df = pd.DataFrame(data)
print(df)
```

```python
x = df[['Hours']]
y = df['marks']
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.5, random_state=42)
print(f"Training set size: {x_train.shape[0]}")
print(f"Testing set size: {x_test.shape[0]}")
```

```python
model = LinearRegression()
model.fit(x_train, y_train)

print(f"Intercept: {model.intercept_}")
print(f"Coefficient: {model.coef_}")
```

```python
y_pred = model.predict(x_test)

predictions = pd.DataFrame({'Actual': y_test, 'Predicted': y_pred})
print(predictions.head())
```

```python
plt.scatter(x_test, y_test, color='blue')
plt.plot(x_test, y_pred, color='red')
plt.xlabel('Hours')
plt.ylabel('Marks')
plt.title('Hours vs Marks')

plt.show()
```

```python
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f"Mean Squared Error: {mse}")
print(f"R-squared: {r2}")
```

```python
# Custom prediction
custom_x = float(input("Enter hours studied: "))

# Create a DataFrame with the same column name used during training
custom_df = pd.DataFrame([[custom_x]], columns=["Hours"])

custom_y = model.predict(custom_df)

print(f"Predicted marks for {custom_x} hours studied: {custom_y[0]:.2f}")
```
